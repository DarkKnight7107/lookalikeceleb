```python
import requests
from bs4 import BeautifulSoup
import urllib.request
import os
import json
```

```python
url = "https://www.imdb.com/list/ls058011111/?sort=list_order%2Casc&page="

headers = {
    'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/110.0.0.0 Safari/537.36'}

persons = []
srcs = []
```

```python
for i in range(1,5):

    response = requests.get(url+str(i), headers=headers)

    html = response.text
    soup = BeautifulSoup(response.text, 'html.parser')
    with open("html.txt",'w') as fh:
        fh.write(html)

    title = soup.find('title').text
    print(f"Page Title: {title}")

    # Find all <script> tags
    script_tags = soup.find_all("script", type="application/ld+json")

    # Extract JSON-LD data
    for script in script_tags:
        try:
            data = json.loads(script.string)  # Parse JSON
            for item in data['itemListElement']:
                persons.append(item['item']['name'])
        except (json.JSONDecodeError, TypeError):
            pass  # Ignore non-JSON scripts

    script_tags = soup.find_all("script", id="__NEXT_DATA__")
    for script in script_tags:
        try:
            data = json.loads(script.string)['props']['pageProps']['mainColumnData']['list']['nameListItemSearch']['edges']     #data['props']['pageProps']['mainColumnData']['list']['primaryImage']['image']['url']
            for d in data:                                                                                                      #data['props']['pageProps']['mainColumnData']['list']['nameListItemSearch']['edges'][0]['listItem']['primaryImage']['url']
                srcs.append(d['listItem']['primaryImage']['url'])
        except (json.JSONDecodeError, TypeError):
            pass  # Ignore non-JSON scripts
```

```python
for i in range(len(persons)):
    filename = "images/" + persons[i] + ".jpg"
    exists = os.path.isfile(filename)
    if exists:
        print("File", filename, "already exists")
    else:
        print("Downloading: ", filename)
        urllib.request.urlretrieve(srcs[i], filename)
```
