```python
import requests
from bs4 import BeautifulSoup
import re
import urllib.request
import os
import json
```

```python
url = "https://www.imdb.com/list/ls058011111/?sort=list_order%2Casc&page=1"
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/110.0.0.0 Safari/537.36'}

response = requests.get(url, headers=headers)
```

```python
html = response.text
soup = BeautifulSoup(response.text, 'html.parser')
with open("test.txt",'w') as fh:
    fh.write(html)
```

```python
title = soup.find('title').text
print(f"Page Title: {title}")
```

```python
imgs = soup.find_all('img', class_ = "ipc-image")
h3s = soup.find_all('h3', class_ = "ipc-title__text")
```

```python
srcs = []
for img in imgs[0:25]:
    srcs.append(img.get('src'))
```

```python
names = []
for h3 in h3s[0:25]:
    names += re.findall('[-a-zA-Z]+[-a-zA-Z ]*',h3.text)
```

```python
# Find all <script> tags
script_tags = soup.find_all("script", type="application/ld+json")
persons = []
# Extract JSON-LD data
for script in script_tags:
    try:
        data = json.loads(script.string)  # Parse JSON
        for item in data['itemListElement']:
            persons.append(item['item']['name'])
    except (json.JSONDecodeError, TypeError):
        pass  # Ignore non-JSON scripts
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager

# Set up Selenium WebDriver
options = Options()
options.headless = True  # Run in headless mode (no GUI)
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# IMDb List URL
url = "https://www.imdb.com/list/ls058011111/?sort=list_order%2Casc&page=1"
driver.get(url)


```

```python
# Extract actor names and image links
actors = driver.find_elements(By.XPATH, "//h3[@class='lister-item-header']/a")
images = driver.find_elements(By.XPATH, "//div[@class='lister-item-image']/a/img")


print(actors)
print(images)
# Store data in a list
actor_data = []

for actor, image in zip(actors, images):
    name = actor.text
    img_src = image.get_attribute("src")  # Get the image source URL
    actor_data.append({"name": name, "image": img_src})

# Print extracted data
for data in actor_data:
    print(f"Name: {data['name']}, Image URL: {data['image']}")

# Close the browser
# driver.quit()
```
